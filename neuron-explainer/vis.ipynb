```python
import json
```

```python
import matplotlib.pyplot as plt
```

```python
data = []
with open(
    "../../../activations/scores/slice_processed_26.jsonl", "r", encoding="utf-8"
) as f:
    for line in f:
        data.append(json.loads(line))
```

```python
scores = [item["score"] for item in data if "score" in item]

plt.figure(figsize=(6, 4))
plt.hist(scores, bins=40, color="steelblue", edgecolor="black")
plt.xlabel("score")
plt.ylabel("count")
plt.title("Experts' score distribution")
plt.tight_layout()
plt.show()
```
